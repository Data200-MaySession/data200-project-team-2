# Week 7: Application Development
This notebook demonstrates a simple Python application that allows users to input academic metrics and receive predicted exam scores using a trained linear regression model.

## 1. Load Libraries and Data

```
# Imports
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import numpy as np

# Load dataset
df = pd.read_csv('updated required data for the project.csv')
df['Attendance'] = df['Attendance'].str.rstrip('%').astype(float)
df.head()
```

## 2. Train Model

```
# Prepare data
X = df[['Study Hours', 'Sleep Duration', 'Attendance']]
y = df['Exam Score']

# Train model
model = LinearRegression()
model.fit(X, y)
print('Model trained successfully.')
```

## 3. Simple Application Interface
Enter your inputs to get the predicted exam score.

```
# User Input Section
def predict_exam_score(study_hours, sleep_duration, attendance):
    input_data = pd.DataFrame([[study_hours, sleep_duration, attendance]],
                              columns=['Study Hours', 'Sleep Duration', 'Attendance'])
    prediction = model.predict(input_data)
    return prediction[0]

# Example Usage
study_hours = float(input('Enter Study Hours: '))
sleep_duration = float(input('Enter Sleep Duration (hrs): '))
attendance = float(input('Enter Attendance (%): '))

predicted_score = predict_exam_score(study_hours, sleep_duration, attendance)
print(f'Predicted Exam Score: {predicted_score:.2f}')
```
