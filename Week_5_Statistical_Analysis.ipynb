# Week 5: Statistical Analysis and Validation
This notebook performs descriptive and inferential statistical analysis on the dataset collected in Week 4, including diagnostics, hypothesis validation, and possible sampling recommendations.

```
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import scipy.stats as stats

# Load the dataset
df = pd.read_csv('/mnt/data/updated required data for the project.csv')

# Convert Attendance to numeric
df['Attendance'] = df['Attendance'].str.replace('%', '').astype(float)

# Preview the data
df.head()
```

## 1. Descriptive Statistics

```
# Summary statistics
df.describe()
```

### Histograms and Pairplots

```
sns.pairplot(df)
plt.show()
```

### Correlation Matrix

```
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

## 2. Inferential Statistics

```
# Pearson correlation between study hours and exam score
corr, p_value = stats.pearsonr(df['Study Hours'], df['Exam Score'])
print(f'Correlation: {corr}, P-value: {p_value}')
```

### Hypothesis Test:
H0: There is no correlation between study hours and exam scores.
H1: There is a correlation.

```
alpha = 0.05
if p_value < alpha:
    print('Reject null hypothesis: Significant correlation found.')
else:
    print('Fail to reject null hypothesis: No significant correlation.')
```

## 3. Diagnostic Measures

```
import statsmodels.api as sm

X = df[['Study Hours', 'Sleep Duration', 'Attendance']]
y = df['Exam Score']
X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
print(model.summary())
```
